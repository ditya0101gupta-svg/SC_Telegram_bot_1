# **LLM**

```python
#LLM

import os
from dotenv import load_dotenv
load_dotenv()

from langchain.chat_models import init_chat_model

#Groq
groq_llm = init_chat_model(
    model="openai/gpt-oss-20b",
    model_provider="groq"
)

groq_llm.invoke("What is a tiger?")
```

```python
from langchain.chat_models import init_chat_model

#GEMINI
gemini = init_chat_model(
    model="gemini-3.1-flash-lite",
    model_provider="google_genai",
    # api_key = os.environ.get('GOOGLE_API_KEY')
)

gemini.invoke("What is a lion?")
```

# **TOOL**

```python
from langchain_core.tools import tool

@tool
def live_cricket_score(country1:str, country2:str)->str:
    """This tool returns the live cricket score between any 2 given countries"""
    return f"Current score between {country1} and {country2} is 120-3"
```

# **AGENT**

```python
from langchain.agents import create_agent

agent = create_agent(
    model=groq_llm,
    system_prompt="""You are a helpful assistant.

For every user question, first call the live_cricket_score tool
using country1="India" and country2="Pakistan".

After receiving the tool result, answer the user's question normally.
""",
    tools=[live_cricket_score]
)

agent.invoke({
    "messages": [
        {"role": "user", "content": "What is a cat?"}
    ]
})
```
